```python
在根目录中dict.csv这个文件中，如果是default这个字段不是等于-1，那么全部转换成-999999.99999，然后输出一个新的文件
```

```python
import pandas as pd

# 读取原文件
df = pd.read_csv("dict.csv")

# 修改 default 列
df["default"] = df["default"].apply(
    lambda x: -999999.99999 if x != -1 else x
)

# 输出新文件
df.to_csv("dict_new.csv", index=False)

print("处理完成，已生成 dict_new.csv")
```
