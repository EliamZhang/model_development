```python
# coding: utf-8
import os, gc, glob, json, logging
import pandas as pd
import numpy as np
from multiprocessing import Pool, cpu_count
from tqdm import tqdm
from collections import defaultdict

# ===============================
# ⭐ 模式切换（只改这里）
# ===============================
RUN_MODE = "feature"     # "score" | "feature"

from v1_0_20251201.run_model_feature import (
    generate_score,
    run_model_feature
)

# =====================================================
# 1. 日志控制
# =====================================================
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)
logger = logging.getLogger(__name__)
logging.getLogger("urllib3").setLevel(logging.WARNING)
logging.getLogger("matplotlib").setLevel(logging.WARNING)

# =====================================================
# 🔧 统一任务参数（只改这里）
# =====================================================
TASK_NAME = "old_reloan_intent_1222"
BASE_DIR = "/opt/workspace/aus_group_drive/Eliam/cdaml_bigq"

# =====================================================
# 2. 路径配置
# =====================================================
SAMPLE_PATH = f"{BASE_DIR}/data/{TASK_NAME}/samples.csv"
TXN_DIR = f"{BASE_DIR}/modeling/tmp_eliam_{TASK_NAME}_variable_illion_transaction_middle"
BAL_DIR = f"{BASE_DIR}/modeling/tmp_eliam_{TASK_NAME}_balance_time_series_middle"
OUT_PATH = f"{BASE_DIR}/data/{TASK_NAME}/sample_with_feature.csv"
ERROR_PATH = f"{BASE_DIR}/data/{TASK_NAME}/sample_with_feature_error_detail.csv"

CHUNK_SIZE = 500_000

# =====================================================
# 3. 工具函数
# =====================================================
def nan_to_empty(x):
    return "" if pd.isna(x) else x


def is_valid_float(x):
    try:
        float(x)
        return True
    except Exception:
        return False


def is_valid_balance_record(rec: dict) -> bool:
    return is_valid_float(rec.get("balance"))

# =====================================================
# 4. 索引构建
# =====================================================
def build_csv_index(csv_dir):
    logger.info(f"📚 构建索引：{csv_dir}")
    index = defaultdict(set)

    for fp in glob.glob(os.path.join(csv_dir, "*.csv")):
        try:
            for chunk in pd.read_csv(
                fp,
                usecols=["user_id", "application_id"],
                chunksize=CHUNK_SIZE
            ):
                for k in zip(chunk.user_id, chunk.application_id):
                    index[k].add(fp)
        except Exception as e:
            logger.warning(f"⚠️ 索引失败跳过：{fp} | {e}")

    logger.info(f"✅ 索引完成：{len(index)} keys")
    return index

# =====================================================
# 5. 读取匹配行
# =====================================================
def load_match_rows_from_index(csv_index, csv_dir, user_id, application_id):
    fps = csv_index.get((user_id, application_id))
    if not fps:
        return pd.DataFrame()

    dfs = []
    for fp in fps:
        try:
            df = pd.read_csv(fp)
            m = df[
                (df["user_id"] == user_id) &
                (df["application_id"] == application_id)
            ]
            if not m.empty:
                dfs.append(m)
        except Exception:
            continue

    return pd.concat(dfs, ignore_index=True) if dfs else pd.DataFrame()

# =====================================================
# 6. 构造模型输入
# =====================================================
def build_input_data(r, txn_index, bal_index):
    uid, aid = r["user_id"], r["application_id"]
    ft = str(r["sample_datetime"])

    txn_df = load_match_rows_from_index(txn_index, TXN_DIR, uid, aid)
    bal_df = load_match_rows_from_index(bal_index, BAL_DIR, uid, aid)

    txn_records = (
        txn_df[
            [
                "amount", "balance", "bank_account_id", "category",
                "dr_cr", "illion_trx_uuid", "text", "third_party",
                "transaction_date", "transaction_id", "trx_type"
            ]
        ].applymap(nan_to_empty).to_dict("records")
        if not txn_df.empty else []
    )

    bal_records = []
    if not bal_df.empty:
        raw_bal = (
            bal_df[
                ["balance", "balance_date", "balance_id", "bank_account_id"]
            ].applymap(nan_to_empty).to_dict("records")
        )
        for rec in raw_bal:
            if is_valid_balance_record(rec):
                bal_records.append(rec)

    return {
        "userId": uid,
        "applicationId": aid,
        "flowTime": ft,
        "illion_raw_transactions": txn_records,
        "illion_day_end_balances": bal_records
    }

# =====================================================
# 7. 单样本执行
# =====================================================
def process_one_sample(args):
    r, txn_index, bal_index = args
    try:
        input_data = build_input_data(r, txn_index, bal_index)

        # ===============================
        # ⭐ 根据模式调用
        # ===============================
        if RUN_MODE == "feature":
            res = run_model_feature(input_vars=input_data)
        else:
            res = generate_score(input_vars=input_data)

        out = dict(r)

        # ===============================
        # ⭐ 统一展开返回结果
        # ===============================
        if isinstance(res, dict):
            # feature 模式：{"features": {...}}
            if RUN_MODE == "feature" and "features" in res:
                if isinstance(res["features"], dict):
                    out.update(res["features"])
            else:
                # score 模式
                for k, v in res.items():
                    if isinstance(v, dict):
                        out.update(v)
                    else:
                        out[k] = v

        return out

    except Exception as e:
        out = dict(r)
        out["feature_error"] = str(e)
        out["_error_detail"] = {
            "user_id": r["user_id"],
            "application_id": r["application_id"],
            "error_message": str(e),
            "input_data_json": json.dumps(
                build_input_data(r, txn_index, bal_index),
                ensure_ascii=False
            )
        }
        return out

    finally:
        gc.collect()

# =====================================================
# 8. 主流程
# =====================================================
def main():
    logger.info("🚀 加载 samples")
    df = pd.read_csv(SAMPLE_PATH)
    df["sample_datetime"] = df["sample_datetime"].astype(str)

    done_keys = set()
    if os.path.exists(OUT_PATH):
        logger.info("🔁 启用断点续跑")
        done_df = pd.read_csv(OUT_PATH, usecols=["user_id", "application_id"])
        done_keys = set(zip(done_df.user_id, done_df.application_id))

    records = [
        r for r in df.to_dict("records")
        if (r["user_id"], r["application_id"]) not in done_keys
    ]

    if not records:
        logger.info("✅ 无待跑样本")
        return

    txn_index = build_csv_index(TXN_DIR)
    bal_index = build_csv_index(BAL_DIR)

    n_workers = max(cpu_count() - 4, 1)
    write_header = not os.path.exists(OUT_PATH)
    write_err_header = not os.path.exists(ERROR_PATH)

    with open(OUT_PATH, "a", encoding="utf-8", newline="") as fout, \
         open(ERROR_PATH, "a", encoding="utf-8", newline="") as ferr:

        with Pool(n_workers) as pool:
            for r in tqdm(
                pool.imap_unordered(
                    process_one_sample,
                    [(r, txn_index, bal_index) for r in records]
                ),
                total=len(records),
                desc="Processing samples"
            ):
                error_detail = r.pop("_error_detail", None)

                pd.DataFrame([r]).to_csv(
                    fout,
                    header=write_header,
                    index=False
                )
                write_header = False
                fout.flush()

                if error_detail:
                    pd.DataFrame([error_detail]).to_csv(
                        ferr,
                        header=write_err_header,
                        index=False
                    )
                    write_err_header = False
                    ferr.flush()

    logger.info(f"✅ 完成：{OUT_PATH}")
    logger.info(f"⚠️ 错误明细：{ERROR_PATH}")

if __name__ == "__main__":
    main()
```

```python
# coding: utf-8
import os, gc, glob, json, logging, hashlib, zlib, shutil, tempfile
import pandas as pd
import numpy as np
from multiprocessing import Pool, cpu_count
from tqdm import tqdm

# ===============================
# ⭐ 模式切换（只改这里）
# ===============================
RUN_MODE = "feature"     # "score" | "feature"

from v1_0_20251201.run_model_feature import generate_score, run_model_feature

# =====================================================
# 1. 日志控制
# =====================================================
logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
logger = logging.getLogger(__name__)
logging.getLogger("urllib3").setLevel(logging.WARNING)
logging.getLogger("matplotlib").setLevel(logging.WARNING)

# =====================================================
# 🔧 统一任务参数（只改这里）
# =====================================================
TASK_NAME = "old_reloan_intent_1222"
BASE_DIR = "/opt/workspace/aus_group_drive/Eliam/cdaml_bigq"

# =====================================================
# 2. 路径配置
# =====================================================
SAMPLE_PATH = f"{BASE_DIR}/data/{TASK_NAME}/samples.csv"
TXN_DIR = f"{BASE_DIR}/modeling/tmp_eliam_{TASK_NAME}_variable_illion_transaction_middle"
BAL_DIR = f"{BASE_DIR}/modeling/tmp_eliam_{TASK_NAME}_balance_time_series_middle"
OUT_PATH = f"{BASE_DIR}/data/{TASK_NAME}/sample_with_feature.csv"
ERROR_PATH = f"{BASE_DIR}/data/{TASK_NAME}/sample_with_feature_error_detail.csv"

# =====================================================
# 3. 性能参数（可调但不影响业务逻辑）
# =====================================================
CHUNK_SIZE = 500_000                      # 读取大文件分块
N_BUCKETS = 1024                          # hash 分桶数量（越大单桶越小；1024 对 50GB 内存安全）
BUCKET_BUILD_WORKERS = max(min(cpu_count() - 2, 24), 1)   # 构桶并行
MODEL_WORKERS = max(cpu_count() - 4, 1)   # 跑模型并行
LRU_BUCKET_CACHE = 2                      # 每个进程最多缓存最近 N 个桶（避免内存爆）
CACHE_ROOT = f"{BASE_DIR}/data/{TASK_NAME}/_bucket_cache" # 断点/复用缓存目录

# =====================================================
# 4. 业务字段配置（列裁剪：只读模型需要的列）
# =====================================================
KEY_COLS = ["user_id", "application_id"]

TXN_COLS = KEY_COLS + [
    "amount", "balance", "bank_account_id", "category", "dr_cr",
    "illion_trx_uuid", "text", "third_party", "transaction_date",
    "transaction_id", "trx_type"
]

BAL_COLS = KEY_COLS + ["balance", "balance_date", "balance_id", "bank_account_id"]

# =====================================================
# 5. 工具函数
# =====================================================
def nan_to_empty(x): return "" if pd.isna(x) else x
def is_valid_float(x):
    try:
        float(x); return True
    except Exception:
        return False
def is_valid_balance_record(rec: dict) -> bool:
    return is_valid_float(rec.get("balance"))

def stable_bucket(user_id, application_id, n_buckets: int) -> int:
    s = f"{int(user_id)}|{int(application_id)}".encode("utf-8", "ignore")
    return zlib.crc32(s) % n_buckets

def has_pyarrow() -> bool:
    try:
        import pyarrow  # noqa
        return True
    except Exception:
        return False

def detect_best_format() -> str:
    return "parquet" if has_pyarrow() else "csv"

BEST_FMT = detect_best_format()

def safe_mkdir(p: str):
    os.makedirs(p, exist_ok=True)

def file_list_any(dir_path: str):
    # parquet 优先；否则 csv
    fps = glob.glob(os.path.join(dir_path, "*.parquet"))
    if fps: return ("parquet", sorted(fps))
    fps = glob.glob(os.path.join(dir_path, "*.csv"))
    return ("csv", sorted(fps))

def make_keys_fingerprint(keys_df: pd.DataFrame) -> str:
    # keys 可能很多：不排序（会爆内存/时间），用采样+总量+滚动 hash 做稳定指纹
    # 只要 keys 不变（同一份 samples），指纹就稳定，缓存可复用
    h = hashlib.sha1()
    h.update(str(len(keys_df)).encode())
    # 取固定步长抽样（稳定，且 O(n/step)）
    step = max(len(keys_df) // 200000, 1)
    sub = keys_df.iloc[::step][["user_id", "application_id"]]
    h.update(pd.util.hash_pandas_object(sub, index=False).values.tobytes())
    return h.hexdigest()

# =====================================================
# 6. 分桶缓存：一次扫描（semi-join）+ 分桶写入（Parquet 优先）
# =====================================================
def _read_source_in_chunks(fp: str, fmt: str, usecols: list, chunksize: int):
    if fmt == "parquet":
        # parquet 走一次性读取更快；但如果很大也可改成 row-group 级别（这里先简化）
        df = pd.read_parquet(fp, columns=usecols)
        yield df
    else:
        for ck in pd.read_csv(fp, usecols=usecols, chunksize=chunksize):
            yield ck

def _write_tmp_batch(df: pd.DataFrame, out_dir: str, prefix: str, fmt: str):
    safe_mkdir(out_dir)
    if fmt == "parquet":
        out = os.path.join(out_dir, f"{prefix}_{os.getpid()}_{np.random.randint(1e9)}.parquet")
        df.to_parquet(out, index=False)
    else:
        out = os.path.join(out_dir, f"{prefix}_{os.getpid()}_{np.random.randint(1e9)}.csv")
        df.to_csv(out, index=False)
    return out

def _bucketize_one_file(args):
    fp, src_fmt, cols, key_set, tmp_root, n_buckets = args
    # 每个 worker：按桶收集（matched 通常远小于全量，所以用 list 累积再落盘）
    bucket_buffers = {}
    written = 0
    try:
        for ck in _read_source_in_chunks(fp, src_fmt, cols, CHUNK_SIZE):
            if ck is None or ck.empty: continue
            # semi-join：只保留命中 keys
            # key_str 用于 isin（比 tuple isin 快很多）
            u = ck["user_id"].astype("int64", errors="ignore")
            a = ck["application_id"].astype("int64", errors="ignore")
            key_str = u.astype(str) + "#" + a.astype(str)
            m = key_str.isin(key_set)
            if not m.any(): continue
            hit = ck.loc[m, cols].copy()
            # 分桶（命中行通常不多，python 循环可接受）
            for _, row in hit.iterrows():
                uid = row["user_id"]; aid = row["application_id"]
                b = stable_bucket(uid, aid, n_buckets)
                bucket_buffers.setdefault(b, []).append(row)
            # 控制内存：每累计到一定量就落盘
            total_rows = sum(len(v) for v in bucket_buffers.values())
            if total_rows >= 200000:
                for b, rows in list(bucket_buffers.items()):
                    if not rows: continue
                    dfb = pd.DataFrame(rows)
                    _write_tmp_batch(dfb, os.path.join(tmp_root, f"b={b}"), "part", BEST_FMT)
                    written += len(dfb)
                bucket_buffers.clear()
        # flush
        for b, rows in bucket_buffers.items():
            if not rows: continue
            dfb = pd.DataFrame(rows)
            _write_tmp_batch(dfb, os.path.join(tmp_root, f"b={b}"), "part", BEST_FMT)
            written += len(dfb)
        return (fp, True, written, "")
    except Exception as e:
        return (fp, False, written, str(e))

def _merge_bucket_parts(bucket_tmp_dir: str, out_fp: str, fmt: str, cols: list):
    part_fmt, part_files = file_list_any(bucket_tmp_dir)
    if not part_files:
        # 写一个空文件占位（方便后续直接判断存在）
        empty = pd.DataFrame(columns=cols)
        if fmt == "parquet": empty.to_parquet(out_fp, index=False)
        else: empty.to_csv(out_fp, index=False)
        return 0

    dfs = []
    if part_fmt == "parquet":
        for pf in part_files:
            dfs.append(pd.read_parquet(pf, columns=cols))
    else:
        for pf in part_files:
            dfs.append(pd.read_csv(pf, usecols=cols))
    df = pd.concat(dfs, ignore_index=True) if dfs else pd.DataFrame(columns=cols)

    # 这里不强制去重：保持业务零侵入（同一 key 多行是合理的）
    if fmt == "parquet": df.to_parquet(out_fp, index=False)
    else: df.to_csv(out_fp, index=False)

    # 清 tmp
    shutil.rmtree(bucket_tmp_dir, ignore_errors=True)
    return len(df)

def build_bucket_cache(data_dir: str, cache_dir: str, sample_keys_df: pd.DataFrame, cols: list, n_buckets: int):
    safe_mkdir(cache_dir)
    meta_fp = os.path.join(cache_dir, "_meta.json")

    keys_fp = make_keys_fingerprint(sample_keys_df)
    if os.path.exists(meta_fp):
        try:
            meta = json.load(open(meta_fp, "r", encoding="utf-8"))
            if meta.get("keys_fingerprint") == keys_fp and meta.get("n_buckets") == n_buckets:
                # 复用缓存：桶文件齐全则直接用（断点续跑友好）
                ok = True
                for b in range(n_buckets):
                    if not (os.path.exists(os.path.join(cache_dir, f"bucket_{b}.parquet")) or
                            os.path.exists(os.path.join(cache_dir, f"bucket_{b}.csv"))):
                        ok = False
                        break
                if ok:
                    logger.info(f"✅ 复用缓存：{cache_dir}（keys 未变）")
                    return
        except Exception:
            pass

    # 重建缓存
    logger.info(f"🧱 构建分桶缓存：{data_dir} -> {cache_dir} | fmt={BEST_FMT} | buckets={n_buckets}")
    for b in range(n_buckets):
        for ext in ("parquet", "csv"):
            fp = os.path.join(cache_dir, f"bucket_{b}.{ext}")
            if os.path.exists(fp): os.remove(fp)

    src_fmt, fps = file_list_any(data_dir)
    if not fps:
        logger.warning(f"⚠️ 目录无数据文件：{data_dir}")
        # 写 meta，避免反复重试
        json.dump({"keys_fingerprint": keys_fp, "n_buckets": n_buckets, "src_fmt": src_fmt}, open(meta_fp, "w", encoding="utf-8"))
        return

    # semi-join key_set（字符串哈希集合）
    kdf = sample_keys_df[["user_id", "application_id"]].copy()
    kdf["user_id"] = pd.to_numeric(kdf["user_id"], errors="coerce").fillna(-1).astype("int64")
    kdf["application_id"] = pd.to_numeric(kdf["application_id"], errors="coerce").fillna(-1).astype("int64")
    key_set = set((kdf["user_id"].astype(str) + "#" + kdf["application_id"].astype(str)).tolist())

    tmp_root = tempfile.mkdtemp(prefix="bucket_tmp_", dir=cache_dir)
    logger.info(f"📦 临时目录：{tmp_root}")

    tasks = [(fp, src_fmt, cols, key_set, tmp_root, n_buckets) for fp in fps]
    bad = 0
    total_written = 0

    with Pool(BUCKET_BUILD_WORKERS) as pool:
        for fp, ok, written, err in tqdm(pool.imap_unordered(_bucketize_one_file, tasks), total=len(tasks), desc="Bucketizing"):
            total_written += int(written)
            if not ok:
                bad += 1
                logger.warning(f"⚠️ 分桶失败：{fp} | {err}")

    # merge 每个桶的 parts -> 最终 bucket 文件（parquet 优先）
    logger.info("🧩 合并桶文件...")
    for b in tqdm(range(n_buckets), desc="Merging buckets"):
        bucket_tmp_dir = os.path.join(tmp_root, f"b={b}")
        out_fp = os.path.join(cache_dir, f"bucket_{b}.{BEST_FMT}")
        _merge_bucket_parts(bucket_tmp_dir, out_fp, BEST_FMT, cols)

    # 清理临时根目录
    shutil.rmtree(tmp_root, ignore_errors=True)

    json.dump(
        {"keys_fingerprint": keys_fp, "n_buckets": n_buckets, "src_fmt": src_fmt, "best_fmt": BEST_FMT, "bad_files": bad, "written_rows": total_written},
        open(meta_fp, "w", encoding="utf-8"),
        ensure_ascii=False, indent=2
    )
    logger.info(f"✅ 缓存构建完成：{cache_dir} | bad_files={bad} | written_rows={total_written}")

# =====================================================
# 7. BucketStore：单样本只读一个桶（带 LRU 小缓存，避免内存爆）
# =====================================================
class BucketStore:
    def __init__(self, cache_dir: str, n_buckets: int, cols: list):
        self.cache_dir = cache_dir
        self.n_buckets = n_buckets
        self.cols = cols
        self._lru = []       # [bucket_id]
        self._cache = {}     # bucket_id -> df

    def _bucket_fp(self, b: int):
        fp_parq = os.path.join(self.cache_dir, f"bucket_{b}.parquet")
        fp_csv = os.path.join(self.cache_dir, f"bucket_{b}.csv")
        if os.path.exists(fp_parq): return ("parquet", fp_parq)
        if os.path.exists(fp_csv): return ("csv", fp_csv)
        # 理论上 build 时就会写占位空文件，这里兜底
        return ("csv", fp_csv)

    def _load_bucket_df(self, b: int) -> pd.DataFrame:
        if b in self._cache:
            # LRU 更新
            if b in self._lru: self._lru.remove(b)
            self._lru.insert(0, b)
            return self._cache[b]

        fmt, fp = self._bucket_fp(b)
        if not os.path.exists(fp):
            df = pd.DataFrame(columns=self.cols)
        else:
            if fmt == "parquet": df = pd.read_parquet(fp, columns=self.cols)
            else: df = pd.read_csv(fp, usecols=self.cols)

        # 放入 cache + LRU 淘汰
        self._cache[b] = df
        self._lru.insert(0, b)
        while len(self._lru) > LRU_BUCKET_CACHE:
            old = self._lru.pop()
            self._cache.pop(old, None)
        return df

    def fetch(self, user_id, application_id) -> pd.DataFrame:
        b = stable_bucket(user_id, application_id, self.n_buckets)
        dfb = self._load_bucket_df(b)
        if dfb.empty: return dfb
        # 这里必须过滤 key（同桶还有别的 key）
        return dfb[(dfb["user_id"] == user_id) & (dfb["application_id"] == application_id)]

# =====================================================
# 8. 构造模型输入（业务零侵入）
# =====================================================
def build_input_data(r, txn_store: BucketStore, bal_store: BucketStore):
    uid, aid = r["user_id"], r["application_id"]
    ft = str(r["sample_datetime"])

    txn_df = txn_store.fetch(uid, aid)
    bal_df = bal_store.fetch(uid, aid)

    txn_records = (
        txn_df[
            ["amount","balance","bank_account_id","category","dr_cr","illion_trx_uuid","text","third_party","transaction_date","transaction_id","trx_type"]
        ].applymap(nan_to_empty).to_dict("records")
        if txn_df is not None and not txn_df.empty else []
    )

    bal_records = []
    if bal_df is not None and not bal_df.empty:
        raw_bal = bal_df[["balance","balance_date","balance_id","bank_account_id"]].applymap(nan_to_empty).to_dict("records")
        for rec in raw_bal:
            if is_valid_balance_record(rec): bal_records.append(rec)

    return {"userId": uid, "applicationId": aid, "flowTime": ft, "illion_raw_transactions": txn_records, "illion_day_end_balances": bal_records}

# =====================================================
# 9. 多进程跑模型（全局 store，避免反复序列化大对象）
# =====================================================
_G_TXN_STORE = None
_G_BAL_STORE = None

def _init_worker(txn_cache_dir, bal_cache_dir, n_buckets):
    global _G_TXN_STORE, _G_BAL_STORE
    _G_TXN_STORE = BucketStore(txn_cache_dir, n_buckets, TXN_COLS)
    _G_BAL_STORE = BucketStore(bal_cache_dir, n_buckets, BAL_COLS)

def process_one_sample(r):
    global _G_TXN_STORE, _G_BAL_STORE
    try:
        input_data = build_input_data(r, _G_TXN_STORE, _G_BAL_STORE)
        if RUN_MODE == "feature":
            res = run_model_feature(input_vars=input_data)
        else:
            res = generate_score(input_vars=input_data)

        out = dict(r)
        if isinstance(res, dict):
            if RUN_MODE == "feature" and "features" in res and isinstance(res.get("features"), dict):
                out.update(res["features"])
            else:
                for k, v in res.items():
                    if isinstance(v, dict): out.update(v)
                    else: out[k] = v
        return out
    except Exception as e:
        out = dict(r)
        out["feature_error"] = str(e)
        try:
            in_json = json.dumps(build_input_data(r, _G_TXN_STORE, _G_BAL_STORE), ensure_ascii=False, default=str)
        except Exception:
            in_json = "{}"
        out["_error_detail"] = {"user_id": r["user_id"], "application_id": r["application_id"], "error_message": str(e), "input_data_json": in_json}
        return out
    finally:
        gc.collect()

# =====================================================
# 10. 主流程
# =====================================================
def main():
    logger.info("🚀 加载 samples")
    df = pd.read_csv(SAMPLE_PATH)
    df["sample_datetime"] = df["sample_datetime"].astype(str)
    df["user_id"] = pd.to_numeric(df["user_id"], errors="coerce").fillna(-1).astype("int64")
    df["application_id"] = pd.to_numeric(df["application_id"], errors="coerce").fillna(-1).astype("int64")

    done_keys = set()
    if os.path.exists(OUT_PATH):
        logger.info("🔁 启用断点续跑（OUT 已存在）")
        done_df = pd.read_csv(OUT_PATH, usecols=["user_id","application_id"])
        done_df["user_id"] = pd.to_numeric(done_df["user_id"], errors="coerce").fillna(-1).astype("int64")
        done_df["application_id"] = pd.to_numeric(done_df["application_id"], errors="coerce").fillna(-1).astype("int64")
        done_keys = set(zip(done_df.user_id, done_df.application_id))

    records = [r for r in df.to_dict("records") if (r["user_id"], r["application_id"]) not in done_keys]
    if not records:
        logger.info("✅ 无待跑样本"); return

    # 1) 一次扫描（semi-join）+ 2) 分桶缓存（hash bucket）+ 3) 列裁剪/分块读取 + 4) Parquet 优先 + 5) 缓存复用
    txn_cache_dir = os.path.join(CACHE_ROOT, f"txn_buckets_{N_BUCKETS}")
    bal_cache_dir = os.path.join(CACHE_ROOT, f"bal_buckets_{N_BUCKETS}")
    safe_mkdir(CACHE_ROOT)

    build_bucket_cache(TXN_DIR, txn_cache_dir, df[["user_id","application_id"]], TXN_COLS, N_BUCKETS)
    build_bucket_cache(BAL_DIR, bal_cache_dir, df[["user_id","application_id"]], BAL_COLS, N_BUCKETS)

    write_header = not os.path.exists(OUT_PATH)
    write_err_header = not os.path.exists(ERROR_PATH)

    with open(OUT_PATH, "a", encoding="utf-8", newline="") as fout, open(ERROR_PATH, "a", encoding="utf-8", newline="") as ferr:
        with Pool(MODEL_WORKERS, initializer=_init_worker, initargs=(txn_cache_dir, bal_cache_dir, N_BUCKETS)) as pool:
            for out in tqdm(pool.imap_unordered(process_one_sample, records), total=len(records), desc="Processing samples"):
                err = out.pop("_error_detail", None)

                pd.DataFrame([out]).to_csv(fout, header=write_header, index=False)
                write_header = False
                fout.flush()

                if err:
                    pd.DataFrame([err]).to_csv(ferr, header=write_err_header, index=False)
                    write_err_header = False
                    ferr.flush()

    logger.info(f"✅ 完成：{OUT_PATH}")
    logger.info(f"⚠️ 错误明细：{ERROR_PATH}")
    logger.info(f"📌 缓存目录（可复用/断点续跑友好）：{CACHE_ROOT}")

if __name__ == "__main__":
    main()
```
